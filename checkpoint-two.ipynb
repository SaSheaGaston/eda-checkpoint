## Visualize

Create any visualizations for your EDA here. Make note in the form of code comments of what your thought process is for your visualizations.

```python
# Visualizing changes in average life expectancy over time globally. 

yearly_avg = (
    df.groupby("Year")["Life expectancy "]
      .mean()
      .reset_index()
)

ax = yearly_avg.plot(x = "Year", y = "Life expectancy ", kind = "line")

ax.set_xlim(2000, 2015)
ax.set_xticks(range(2000, 2016))
ax.set_xlabel("Year")
ax.set_ylabel("Life Expectancy in Years")
plt.xticks(rotation=45)
plt.title("Average Global Life Expectancy 2000 - 2015 (133 Countries)")

plt.tight_layout()
plt.show()



```

```python
# Visualizing changes in life expectancy over time, comparing the trend for developing vs. developed countries. 


year_status_avg = (
    df.groupby(["Year", "Status"])["Life expectancy "]
      .mean()
      .reset_index()
)

plot_df = year_status_avg.pivot(
    index="Year",
    columns="Status",
    values="Life expectancy "
)

ax = plot_df.plot(
    kind="line",
    color=["tab:blue", "tab:orange"]
)

ax.set_xlim(2000, 2015)
ax.set_xticks(range(2000, 2016))
ax.tick_params(axis = "x", rotation=45)

ax.set_xlabel("Year")
ax.set_ylabel("Life Expectancy in Years")
ax.legend(title="Country Status")
plt.title("Average Life Expectancy Over Time: Developing vs Developed Countries")

plt.tight_layout()
plt.show()
```

```python
# Visualizing average global adult mortality rate trends over time.

yearly_avg = (
    df.groupby("Year")["Adult Mortality"]
      .mean()
      .reset_index()
)


ax = yearly_avg.plot(
    x = "Year",
    y = "Adult Mortality",
    kind = "line"
)

ax.set_xlim(2000, 2015)
ax.set_xticks(range(2000, 2016))
ax.set_xlabel("Year")
ax.set_ylabel("Average Adult Mortality Rate per 1,000 Population")
plt.xticks(rotation = 45)
plt.title("Average Global Adult Mortality Rate per 1,000 Population (Ages 15–60)")

plt.tight_layout()
plt.show()
```

```python
# Visualizing average adult mortality rates, developing vs. developed countries.

year_status_avg_mortality = (
    df.groupby(["Year", "Status"])["Adult Mortality"]
      .mean()
      .reset_index()
)


plot_mortality = year_status_avg_mortality.pivot(
    index = "Year",
    columns = "Status",
    values = "Adult Mortality"
)

ax = plot_mortality.plot(
    kind = "line",
    color = ["tab:blue", "tab:orange"]
)

ax.set_xlim(2000, 2015)
ax.set_xticks(range(2000, 2016))
ax.tick_params(axis = "x", rotation = 45)

ax.set_xlabel("Year")
ax.set_ylabel("Average Adult Mortality Per 1,000 Population")
ax.legend(title = "Country Status")
plt.title("Average Global Adult Mortality Rate: Developing vs Developed Countries")

plt.tight_layout()
plt.show()
```

```python
# Visualizing average infant deaths, developing vs. developed countries.

df_filtered_infant = df[df["infant deaths"] <= 1000]

year_status_avg_infants = (df_filtered_infant.groupby(["Year", "Status"])["infant deaths"].mean().reset_index())


plot_infants = year_status_avg_infants.pivot(
    index = "Year",
    columns = "Status",
    values = "infant deaths"
)

ax = plot_infants.plot(
    kind = "line",
    color = ["tab:blue", "tab:orange"]
)

ax.set_xlim(2000, 2015)
ax.set_xticks(range(2000, 2016))
ax.tick_params(axis = "x", rotation = 45)

ax.set_xlabel("Year")
ax.set_ylabel("Average Infant Deaths Per 1,000 Live Births")
ax.legend(title = "Country Status")
plt.title("Average Number of Infant Deaths: Developing vs Developed Countries")

plt.tight_layout()
plt.show()
```

```python
# Correlation between GDP and infant deaths using log scale 

df_filtered_infant = df[df["infant deaths"] <= 1000]



x = df_filtered_infant["GDP"]
y = df_filtered_infant["infant deaths"]


m, b = np.polyfit(np.log10(x), y, 1)

plt.figure(figsize=(8, 6))


plt.scatter(x, y, alpha=0.4, label="Data Points")


plt.xscale('log')

# to create evenly spaced points along the log scale
x_range = np.geomspace(x.min(), x.max(), 100)
y_trend = m * np.log10(x_range) + b
plt.plot(x_range, y_trend, color="red", linewidth=2, label="Log-Linear Trend")


plt.xlabel("Gross Domestic Product (GDP) - Log Scale ($)")
plt.ylabel("Infant Deaths Per 1,000 Live Births")
plt.title("Relationship Between GDP and Infant Deaths")
plt.legend()
plt.grid(True, which="both", ls="-", alpha=0.2) 
plt.show()
```
