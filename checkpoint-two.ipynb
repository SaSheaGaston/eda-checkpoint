## Visualize

Create any visualizations for your EDA here. Make note in the form of code comments of what your thought process is for your visualizations.

```python
# Visualizing changes in average life expectancy over time globally. 

yearly_avg = (
    df.groupby("Year")["Life expectancy "]
      .mean()
      .reset_index()
)

ax = yearly_avg.plot(x = "Year", y = "Life expectancy ", kind = "line")

ax.set_xlim(2000, 2015)
ax.set_xticks(range(2000, 2016))
ax.set_xlabel("Year")
ax.set_ylabel("Life Expectancy in Years")
plt.xticks(rotation=45)
plt.title("Average Global Life Expectancy 2000 - 2015 (133 Countries)")

plt.tight_layout()
plt.show()



```

```python
# Visualizing changes in life expectancy over time, comparing the trend for developing vs. developed countries. 


year_status_avg = (
    df.groupby(["Year", "Status"])["Life expectancy "]
      .mean()
      .reset_index()
)

plot_df = year_status_avg.pivot(
    index="Year",
    columns="Status",
    values="Life expectancy "
)

ax = plot_df.plot(
    kind="line",
    color=["tab:blue", "tab:orange"]
)

ax.set_xlim(2000, 2015)
ax.set_xticks(range(2000, 2016))
ax.tick_params(axis = "x", rotation=45)

ax.set_xlabel("Year")
ax.set_ylabel("Life Expectancy in Years")
ax.legend(title="Country Status")
plt.title("Average Life Expectancy Over Time: Developing vs Developed Countries")

plt.tight_layout()
plt.show()
```

```python
# Visualizing average global adult mortality rate trends over time.

yearly_avg = (
    df.groupby("Year")["Adult Mortality"]
      .mean()
      .reset_index()
)


ax = yearly_avg.plot(
    x = "Year",
    y = "Adult Mortality",
    kind = "line"
)

ax.set_xlim(2000, 2015)
ax.set_xticks(range(2000, 2016))
ax.set_xlabel("Year")
ax.set_ylabel("Average Adult Mortality Rate per 1,000 Population")
plt.xticks(rotation = 45)
plt.title("Average Global Adult Mortality Rate per 1,000 Population (Ages 15–60)")

plt.tight_layout()
plt.show()
```

```python
# Visualizing average adult mortality rates, developing vs. developed countries.

year_status_avg_mortality = (
    df.groupby(["Year", "Status"])["Adult Mortality"]
      .mean()
      .reset_index()
)


plot_mortality = year_status_avg_mortality.pivot(
    index = "Year",
    columns = "Status",
    values = "Adult Mortality"
)

ax = plot_mortality.plot(
    kind = "line",
    color = ["tab:blue", "tab:orange"]
)

ax.set_xlim(2000, 2015)
ax.set_xticks(range(2000, 2016))
ax.tick_params(axis = "x", rotation = 45)

ax.set_xlabel("Year")
ax.set_ylabel("Average Adult Mortality Per 1,000 Population")
ax.legend(title = "Country Status")
plt.title("Average Global Adult Mortality Rate: Developing vs Developed Countries")

plt.tight_layout()
plt.show()
```
